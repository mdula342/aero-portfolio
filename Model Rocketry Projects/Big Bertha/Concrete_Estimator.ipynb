```python
def area(points):

    n = len(points)

    closed_points = points + [points[0]]
    total = 0
    
    for i in range(n):
        x1, y1 = closed_points[i]
        x2, y2 = closed_points[i+1]
        total += x1*y2 - x2*y1

    return abs(total)/2

points = [(0, 0), (10, 0), (10, 2), (0, 2)]

result = area(points)
print(result)
```

```python
def shoelace_area(points):
    n = len(points)
    total = 0

    for i in range(n):
        x1, y1 = points[i]
        x2, y2 = points[(i + 1) % n]
        total += x1 * y2 - x2 * y1

    return abs(total) / 2


points = [(0,0), (4,0), (4,3), (0,3)]

print(shoelace_area(points))
```

```python
import open3d as o3d
import numpy as np

# fake RTK-like points (replace with your real data later)
points = np.array([
    [0, 0, 0],
    [10, 0, 0.2],
    [10, 5, 0.3],
    [0, 5, 0.1]
])

pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)

# STEP 1: triangulate surface (TIN-like)
mesh = o3d.geometry.TriangleMesh.create_from_point_cloud_alpha_shape(
    pcd, alpha=2.0
)

# STEP 2: compute surface area
area = mesh.get_surface_area()

print("Surface area:", area)
```
