```python
import numpy as np

# N = number of trials for each data set
N = 10
# Measurements: Length = Ls, Diameter = Ds, Period = Ps, Mass = Ms, Weight = Ws
Ls = [79.00, 78.09, 79.00, 78.09, 79.10, 78.40, 78.31, 78.61, 78.35, 78.50]
Ds = [6.31, 6.33, 6.25, 6.22, 6.20, 6.24, 6.13, 6.10, 6.15, 6.20]
Ps = [1.80, 1.77, 1.72, 1.87, 1.82, 1.80, 1.83, 1.80, 1.80, 1.86]
Ms = [58.9, 58.8, 58.9, 58.8, 58.9, 58.8, 58.9, 58.9, 58.8, 58.8]
Ws = [0.60, 0.60, 0.60, 0.60, 0.60, 0.59, 0.59, 0.59, 0.59, 0.59]

# Find the mean of each data set
def x_bar(my_list):
    xbar = sum(my_list)/N
    return(xbar)

# List of means
xbars = [78.545, 6.213, 1.807, 58.85, 0.595]
lcm,dcm,Ts,mg,wN = [78.545, 6.213, 1.807, 58.85, 0.595]

# Compute the list of (xsubi - xbar)^2 in the standard deviation sum 
def dvs(mlist, mlistnum):
    dvslist = []
    for i in mlist:
        dvslist.append((i-xbars[mlistnum])**2)
    return dvslist

# List of (xsubi - xbar)^2 for each data set
Lsdvs = dvs(Ls, 0)
Dsdvs = dvs(Ds, 1)
Psdvs = dvs(Ps, 2)
Msdvs = dvs(Ms, 3)
Wsdvs = dvs(Ws, 4)

# Calculate standard deviation for each data set
def stdv(dvslist):
    stdev = (sum(dvslist)/N)**(1/2)
    return stdev

# Standard deviations
Lsstdev = stdv(Lsdvs)
Dsstdev = stdv(Dsdvs)
Psstdev = stdv(Psdvs)
Msstdev = stdv(Msdvs)
Wsstdev = stdv(Wsdvs)

# List of standard deviations
stdevs = [Lsstdev, Dsstdev, Psstdev, Msstdev, Wsstdev]

# List of standard errors of the mean
SEMS = [i/(N**(1/2)) for i in stdevs]

# Calculate gravitational acceleration, g, using units m and s
l = lcm*0.01
T = Ts
g = 4*np.pi**2*(l/T**2)

# Calculate the standard error for g
SEgj = np.sqrt((4*np.pi**2/T**2*SEMS[0]*0.01)**2 + (8*np.pi**2*l/T**3*SEMS[2])**2)

# Calculate discrepancy for g
Discrepancy = np.abs(g-9.81)

def SEg(SEgt, SEgl):
    return np.sqrt(SEgt**2 + SEgl**2)

# Derivative method for SEg - define SET and SEl
SEl = SEMS[0]*0.01
SET = SEMS[2]

SEgtd = 8*np.pi**2*l*SET/T**3
SEgld = 4*np.pi**2*SEl/T**2
SEgd = SEg(SEgtd, SEgld)

# Computational method for SEg

SEgtc = 4*np.pi**2*l*np.abs((1/(T + SET)**2) - (1/T**2))
SEglc = 4*np.pi**2*SEl/T**2
SEgc = SEg(SEgtc, SEglc)


print(xbars)
print(stdevs)
print(SEMS)
print(g, np.abs(g-9.81))
print(SEgd, SEgc, SEgj)
```
